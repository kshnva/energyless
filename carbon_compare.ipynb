Sample experiment to run different simulations for a data center, assuming different geographical locations, by providing different carbon trace files in order to map electricity usage to the respective areas carbon trace

```python
import os
import subprocess
import platform

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt


# Detect OS and set JAVA_HOME accordingly
os_type = platform.system()

if os_type == "Darwin":  # macOS
    os.environ["JAVA_HOME"] = "/opt/homebrew/opt/openjdk@21"
    os.environ["PATH"] = f"{os.environ['JAVA_HOME']}/bin:" + os.environ["PATH"]
elif os_type == "Linux":  # Ubuntu/Linux
    java_home = os.environ.get("JAVA_HOME", "/usr/lib/jvm/java-21-openjdk-amd64")
    os.environ["JAVA_HOME"] = java_home
    os.environ["PATH"] = f"{os.environ['JAVA_HOME']}/bin:" + os.environ["PATH"]
else:
    print(
        f"Warning: Unsupported OS ({os_type}). JAVA_HOME may need to be set manually."
    )

carbon_workflow = f"input/experiments/carbon_workflow_experiment.json"
base_experiment = f"input/experiments/base_experiment.json"


# Run carbon workflow experiment
subprocess.run(
    [
        "OpenDCExperimentRunner/bin/OpenDCExperimentRunner",
        "--experiment-path",
        carbon_workflow,
    ]
)

# Read carbon workflow results
df_powerSource_carbon = pd.read_parquet(
    "output/carbon_workflow_experiment/raw-output/0/seed=0/powerSource.parquet"
)
total_carbon_emissions_carbon_workflow = df_powerSource_carbon["carbon_emission"].sum() / 1000  # Convert to kgCO2
```

```python
# Run base experiment
subprocess.run(
    [
        "OpenDCExperimentRunner/bin/OpenDCExperimentRunner",
        "--experiment-path",
        base_experiment,
    ]
)

# Read base experiment results
df_powerSource_base = pd.read_parquet(
    "output/base_experiment/raw-output/0/seed=0/powerSource.parquet"
)
total_carbon_emissions_base_experiment = df_powerSource_base["carbon_emission"].sum() / 1000  # Convert to kgCO2
```

```python
print(f"Total Carbon Emission (Carbon Workflow): {total_carbon_emissions_carbon_workflow:.2f} kgCO2")
print(f"Total Carbon Emission (Base Experiment): {total_carbon_emissions_base_experiment:.2f} kgCO2")
print(f"\nDifference: {abs(total_carbon_emissions_carbon_workflow - total_carbon_emissions_base_experiment):.2f} kgCO2")
print(f"Reduction: {((total_carbon_emissions_base_experiment - total_carbon_emissions_carbon_workflow) / total_carbon_emissions_base_experiment * 100):.2f}%")
```
