# Weight Experimentation for Workflows

The goal of this notebook will be to figure out the optimal weighter for the workflow schedueler on opendc. The optimial weighers will be weighers that optimize makespan in the experiments


```python
# Import all necessary libraries
import subprocess 
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

# Utility functions

# Run the experiment
def run_experiment(experiment_name):
    pathToScenario = fr"input\experiments\{experiment_name}.json"
    runnerPath = r"OpenDCExperimentRunner\bin\OpenDCExperimentRunner.bat"

    result = subprocess.run(
    f'"{runnerPath}" --experiment-path "{pathToScenario}"',
    shell=True,
    capture_output=True,
    text=True,
    )

    print("STDOUT:\n", result.stdout)
    print("STDERR:\n", result.stderr)
    print("Return code:", result.returncode)

# Load output data from experiment
def load_data(experiment_path):
    df_host = pd.read_parquet(f"output/{experiment_path}/raw-output/0/seed=0/host.parquet")
    df_powerSource = pd.read_parquet(f"output/{experiment_path}/raw-output/0/seed=0/powerSource.parquet")
    df_task = pd.read_parquet(f"output/{experiment_path}/raw-output/0/seed=0/task.parquet")
    df_service = pd.read_parquet(f"output/{experiment_path}/raw-output/0/seed=0/service.parquet")
    return df_host, df_powerSource, df_task, df_service
```

```python
# Run simulation
experiment_name = "emily_workflow_experiment"
run_experiment(experiment_name)
df_host, df_powerSource, df_task, df_service = load_data(experiment_name)

```
